## Reshaping, stacking, squeezing and unsqueezing tensors

- Reshaping : reshapes an input tensor to a defined shape
- Viewing : Return a view of an input tensor of certain shape but keep the same memory as the orignal tensor
- Stacking : combine multiple tensors on top of eah other (vstack) or side by side (hstack)

- Squeeze : remove all `1` dimensions from a tensor
- Unsqueeze : add a `1` dimension to a target tensor

- Permute : return a view of the input with dimensions permuted (swapped) in a certain way

```python
import torch
x = torch.arange(1, 11, 1)
x, x.shape
```

```python
# add extra dim

x_reshaped = x.reshape(5, 2) # 10, 1
x_reshaped
```

```python
# Change the view

z = x.view(1, 10)
z, z.shape
```

If we change z in this view function, the original value x is also changed.

```python
z[:, 0] = 5
x
```

```python
# Stack tensors on top of each other

x_stacked = torch.stack([x, x, x, x])
x_stacked
```

```python
# Squeeze

x = torch.zeros(2, 1, 2, 1, 2)
x.size()
```

```python
y = torch.squeeze(x)
y.size()
```

```python
x_reshaped.size()
```

```python
y = torch.arange(10, 50, 5)
print(f"\nPrevious tensor: {y}")
print(f"Previous shape: {y.size()}\n")

y = y.reshape(1, 8)
print(f"New tensor : {y}")
print(f"New shape: {y.size()}")

print(f"\nAfter squeezing...")
y_squeezed = torch.squeeze(y)

print(f"Updated shape: {y_squeezed.size()}")
```

```python
# Unsqueeze
y_unsqueezed = y_squeezed.unsqueeze(dim=1)

print(f"Before unsqueezing...\nPrevious target: {y_squeezed.size()}\n\nAfter unsqueezing...\nNew Shape: {y_unsqueezed.size()} ")
```

```python
# Permute - rearrange the dimensions of a target tensor in a specified order

x_org = torch.rand(224, 224, 3)
x_permuted = x_org.permute(2, 0, 1)

print(f"Original tensor shape: {x_org.size()}\n\nAfter permuting...\nNew Tensor shape: {x_permuted.size()}")
```

## Reproducibility (Trying to take the random out ot random)

```python
import torch

random_A = torch.rand(3, 4)
random_B = torch.rand(3, 4)

print(random_A)
print(random_B)
print(random_A == random_B)
```

```python
# make random but reproducible tensors

import torch

RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_D = torch.rand(3, 4)

print(random_C)
print(random_D)
print(random_C == random_D)
```
